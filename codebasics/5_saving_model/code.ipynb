```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("canada_per_capita_income.csv")
```

```python
plt.scatter(df['year'], df['per capita income (US$)'], color='red', marker='+')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['year']], df['per capita income (US$)'])
```

```python
reg.predict([[3300]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
m = reg.coef_
x = 3300
b = reg.intercept_

y = m*x+b

print(y)
```

```python
d = pd.read_csv("canada_per_capita_income.csv")
predict_this = d[['year']].tail(3)+3
predict_this
```

```python
p = reg.predict(predict_this)
p
```

```python
predict_this["per capita income (US$)"] = p

df_lengkap = pd.concat([d, predict_this], ignore_index=True)

print(df_lengkap.tail(3))
```

```python
df_lengkap.to_csv("canada_per_capita_income.csv", index=False)
```
