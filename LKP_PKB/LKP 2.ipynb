```python
def m_reg(uts, uas, prak, *keak):
    return (uts*0.3 + uas*0.35
            + prak*0.3 + (sum(keak)/len(keak))*0.05)


def m_mat(uts, uas, prak, *keak, matrikulasi):  # m itu keyword only
    return (uts*0.25 + uas*0.25
            + prak*0.2 + (sum(keak)/len(keak))*0.05
            + matrikulasi*0.25)


raw = input().replace(" ", "").split(",")
m_indicator = False

# cek elemen terakhir ada "=" atau ga, kalo iya brarti keyword only
if "=" in raw[-1]:
    key, value = raw.pop().split("=")
    keyword = {key: int(value)}
    m_indicator = True
else:
    keyword = {}

data = list(map(int, raw))

if (m_indicator):
    print(m_mat(*data, **keyword))
else:
    print(m_reg(*data))
```

```python
import numpy as np

m = np.full((3, 5), "0", dtype=str)
m[[0, 1, 2], [2, 1, 2]] = "1"  # kiri row kanan cols

temp = []  # otw kita flatten
for i, row in enumerate(m):  # enum itu return tuple
    if i % 2 == 0:  # kalo baris genap
        temp.append(row)  # copas aja barisnya
    else:  # sisanya (kalo baris ganjil)
        temp.append(row[::-1])  # slice dari awal sampe akhir, mundur

res = np.concatenate(temp)

for t in range(res.size):
    print(f"t={t+1}")

    res[t] = "X"

    ori = res.reshape(m.shape).copy()  # bikin 2d dulu baru bisa dibalikin

    for r in range(ori.shape[0]):
        if r % 2 == 1:   # baris ganjil
            ori[r] = ori[r][::-1]  # reverse balik biar ke orientasi "asli"

    for row in ori:
        print(*row)

    res[t] = "0"  # jadiin 0
```
