```python
class SimpleKB:
    def __init__(self):
        self.facts = set()

    def tell(self, *facts):
        self.facts.update(facts)

    def ask_if_true(self, fact):
        return fact in self.facts


kb = SimpleKB()

#fakta awal
kb.tell(
    "Agent(1,1)",
    "Totem(2,4)",
    "Wumpus(2,3)",
    "Pit(3,2)",
    "Pit(3,4)",
    "Pit(4,3)"
)

#observasi pas jalan
kb.tell(
    "NoBreeze(1,1)", "NoStench(1,1)",
    "NoBreeze(1,2)", "NoStench(1,2)",
    "NoBreeze(2,1)", "NoStench(2,1)",
    "Breeze(2,2)", "Stench(2,2)",
    "NoBreeze(1,3)", "Stench(1,3)",
    "NoBreeze(1,4)", "NoStench(1,4)"
)

#hasil inferensi
kb.tell(
    "Safe(1,2)", "Safe(2,1)",
    "Safe(1,3)", "Safe(2,2)",
    "Safe(3,1)",
    "Wumpus(2,3)",
    "Safe(1,4)",
    "Safe(2,4)"
)

queries = [
    "Safe(1,2)",
    "Safe(2,1)",
    "Safe(2,2)",
    "Wumpus(2,3)",
    "Safe(1,4)",
    "Safe(2,4)",
    "Totem(2,4)"
]

for q in queries:
    print(f"{q:15} -> {kb.ask_if_true(q)}")
```
