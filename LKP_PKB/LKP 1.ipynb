```python
i = int(input())
k = "Total harga yang harus dibayarkan :"
if (i > 50000):
    print(f"{k}{int(i-(i*0.27))}")
elif (30000 <= i <= 50000):
    print(f"{k}{int(i-(i*0.22))}")
else:
    print(f"{k}{int(i)}")
```

```python
i_1 = input()
k = "Anda mendapatkan golongan UKT "
if (i_1 == "BIDIKMISI"):
    print(f"{k}1")
elif (i_1 == "TIDAK BIDIKMISI"):
    i_2 = int(input())
    if (i_2 > 10_000_000):
        print(f"{k}5")
    elif (7_000_000 < i_2 <= 10_000_000):
        print(f"{k}4")
    elif (4_000_000 < i_2 <= 7_000_000):
        print(f"{k}3")
    elif (i_2 <= 4_000_000):
        print(f"{k}2")
```

```python
w = int(input("weight: "))
h = int(input("height: "))
bmi = w/((h/100)**2)
kategori = ("Below normal weight" if bmi < 18.5 else
            "Normal weight" if bmi < 25 else
            "Overweight" if bmi < 30 else
            "Class I Obesity" if bmi < 35 else
            "Class II Obesity" if bmi < 40 else
            "Class III Obesity")

print(f"{kategori}")
```
